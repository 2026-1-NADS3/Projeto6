```python
# TESTE UNITÁRIO 1: Validação de Prontuário

def validar_tamanho_prontuario(texto):
    # A regra de negócio exige no máximo 500 caracteres
    return len(texto) <= 500

print("--- Iniciando Teste Unitário 1 ---")
print("Digite uma anotação para o prontuário do paciente (limite 500 caracteres):")
entrada_usuario = input("> ")

resultado = validar_tamanho_prontuario(entrada_usuario)

print("\n[Resultado do Teste]")
print(f"Tamanho do texto digitado: {len(entrada_usuario)} caracteres.")
if resultado:
    print("✅ Status: PASSOU! O prontuário está dentro do limite e seria salvo.")
else:
    print("❌ Status: FALHOU! O texto excedeu o limite de caracteres.")
```

```python
# TESTE UNITÁRIO 2: Higienização de E-mail

def higienizar_email(email):
    # Converte tudo para minúsculo e remove espaços em branco nas pontas
    return email.lower().strip()

print("\n--- Iniciando Teste Unitário 2 ---")
print("Simule o cadastro de um paciente. Digite um e-mail contendo letras MAIÚSCULAS:")
email_entrada = input("> ")

email_tratado = higienizar_email(email_entrada)

print("\n[Resultado do Teste]")
print(f"E-mail Original: '{email_entrada}'")
print(f"E-mail Tratado pelo Sistema: '{email_tratado}'")
if email_tratado == email_entrada.lower().strip():
    print("✅ Status: PASSOU! O e-mail foi higienizado corretamente.")
else:
    print("❌ Status: FALHOU! Ocorreu um erro na conversão.")
```

```python
# TESTE UNITÁRIO 3

def calcular_valor_pacote(qtd_sessoes, valor_sessao):
    # Regra de negócio: 10 ou mais sessões ganham 10% de desconto
    total = qtd_sessoes * valor_sessao
    if qtd_sessoes >= 10:
        total = total - (total * 0.10)
    return total

print("\n--- Iniciando Teste Unitário 3 ---")
try:
    sessoes_input = int(input("Digite a quantidade de sessões que o paciente vai fechar: "))
    valor_input = float(input("Digite o valor da sessão avulsa (ex: 120.00): "))

    valor_final = calcular_valor_pacote(sessoes_input, valor_input)

    print("\n[Resultado do Teste]")
    print(f"Quantidade: {sessoes_input} sessões | Valor base: R$ {valor_input:.2f}")

    # Calculando na mão para o assert do teste
    valor_esperado = sessoes_input * valor_input
    if sessoes_input >= 10:
        valor_esperado *= 0.90
        print("-> Desconto de 10% aplicado pela regra de negócio!")

    print(f"Valor calculado pelo sistema: R$ {valor_final:.2f}")

    if valor_final == valor_esperado:
        print("✅ Status: PASSOU! O cálculo financeiro está correto.")
    else:
        print("❌ Status: FALHOU! O cálculo não bateu.")
except ValueError:
    print("❌ Erro: Por favor, digite apenas números válidos para testar.")
```

```python
# TESTE UNITÁRIO 4: Validação de Formato de Hora

def validar_formato_hora(horario):
    # Verifica tamanho e se tem os dois pontos
    if len(horario) != 5 or ":" not in horario:
        return False

    partes = horario.split(":")
    # Verifica se conseguimos separar em duas partes numéricas
    if len(partes) != 2 or not partes[0].isdigit() or not partes[1].isdigit():
        return False

    hora = int(partes[0])
    minuto = int(partes[1])

    # Verifica se os números fazem sentido no relógio
    return 0 <= hora <= 23 and 0 <= minuto <= 59

print("\n--- Iniciando Teste Unitário 4 ---")
print("Simulação do campo de horário da agenda.")
horario_entrada = input("Digite o horário da sessão (formato HH:MM): ")

eh_valido = validar_formato_hora(horario_entrada)

print("\n[Resultado do Teste]")
print(f"Horário digitado: '{horario_entrada}'")
if eh_valido:
    print("✅ Status: PASSOU! Formato de hora reconhecido e validado.")
else:
    print("❌ Status: FALHOU! Formato inválido. O sistema deve bloquear o salvamento.")
```

```python
#TESTE DE INTEGRAÇÃO 1: Cadastro + BD

# Módulo 1: Simulação do Banco de Dados
def salvar_paciente_bd(nome, cpf):
    if not nome or not cpf:
        return {"status": "erro", "mensagem": "Dados incompletos recebidos pelo BD."}
    # Simulando a criação de um ID no banco
    import random
    id_gerado = random.randint(1000, 9999)
    return {"status": "sucesso", "id": id_gerado}

# Módulo 2: Controlador de Cadastro (Integra com o BD)
def integrar_cadastro(nome, cpf):
    print("Processando dados e enviando para o banco de dados...")
    resposta = salvar_paciente_bd(nome, cpf)
    return resposta

print("\n--- Iniciando Teste de Integração 1 ---")
nome_paciente = str(input("Digite o nome do paciente: "))
cpf_paciente = int(input("Digite o CPF do paciente: "))

resultado = integrar_cadastro(nome_paciente, cpf_paciente)

print("\n[Resultado da Integração]")
if resultado["status"] == "sucesso":
    print(f"✅ Status: PASSOU! Integração bem sucedida. Paciente salvo com ID: {resultado['id']}")
else:
    print(f"❌ Status: FALHOU! Erro na integração: {resultado['mensagem']}")
```

```python
# TESTE DE INTEGRAÇÃO 2: Agendamento + Agenda

# Módulo 1: Simulação da Agenda da Fisioterapeuta
horarios_ocupados = ["14:00", "15:00", "18:00"]

def verificar_disponibilidade(horario):
    if horario in horarios_ocupados:
        return False
    return True

# Módulo 2: Controlador de Agendamento
def tentar_agendar_sessao(horario):
    print(f"Consultando o módulo de agenda para o horário {horario}...")
    disponivel = verificar_disponibilidade(horario)

    if disponivel:
        # Simulando salvar na agenda
        horarios_ocupados.append(horario)
        return "Sessão agendada com sucesso!"
    else:
        return "Conflito de agenda! Horário já reservado para outro paciente."

print("\n--- Iniciando Teste de Integração 2 ---")
print(f"Horários atualmente ocupados na clínica: {horarios_ocupados}")
horario_desejado = input("Digite um horário para tentar agendar a sessão (ex: 10:00 ou 14:00): ")

resposta_agendamento = tentar_agendar_sessao(horario_desejado)

print("\n[Resultado da Integração]")
print(f"Retorno do Sistema: {resposta_agendamento}")
print(f"Agenda atualizada: {horarios_ocupados}")
```
